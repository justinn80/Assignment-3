```python
"""
Unit 5: Evolving Your IPD Agent with Genetic Algorithms
========================================================

In Unit 4, you hand-coded agent strategies like Tit-for-Tat.
In Unit 5, you'll use evolution to DISCOVER strategies automatically.

By the end, you'll have an evolved agent saved as a JSON file for the tournament!
"""

# ============================================================================
# SETUP: Import everything from previous units
# ============================================================================

import random
import statistics
import numpy as np
import json
import matplotlib.pyplot as plt
from agents import Agent, INVEST, UNDERCUT
from game_engine import Game, Tournament


# Import all hand-coded agents from Unit 4
from agents import (
    AlwaysInvestAgent, AlwaysUndercutAgent, TitForTatAgent,
    GrimTriggerAgent, PavlovAgent, TitForTwoTatsAgent,
    GenerousTitForTatAgent, AdaptiveAgent, RandomAgent,
    SuspiciousTitForTatAgent, GradualAgent, HardMajorityAgent,
    SoftMajorityAgent, ProberAgent
)

print("✅ Imports complete!")
```

```python

# ============================================================================
# PART 1: THE EVOLVABLE AGENT (PROVIDED - Read but don't modify)
# ============================================================================

class EvolvableAgent(Agent):
    """
    An agent whose strategy is controlled by 6 genes (numbers from 0 to 1).
    
    Think of genes as "strategy dials" you can tune:
    
    Gene[0] - Initial Cooperation (0=hostile start, 1=friendly start)
    Gene[1] - Cooperation Response (if opponent cooperates, how likely to cooperate back?)
    Gene[2] - Defection Response (if opponent defects, how likely to defect back?)
    Gene[3] - Forgiveness (after retaliation, chance to forgive)
    Gene[4] - Memory Length (0=last 1 round, 1=last 10 rounds)
    Gene[5] - Retaliation Threshold (what % of defections triggers retaliation)
    """
    
    def __init__(self, genes=None, name="Evolved Agent"):
        if genes is None:
            # Create random genes if none provided
            genes = [random.random() for _ in range(6)]
        
        self.genes = genes
        super().__init__(name, f"Genes: [{', '.join([f'{g:.2f}' for g in genes])}]")
    
    def choose_action(self) -> bool:
        """Decision logic based on genes - SIMPLIFIED FOR LEARNING"""
        
        # First 3 rounds: use initial cooperation gene
        if self.round_num < 3:
            return random.random() < self.genes[0]
        
        # Calculate memory window (gene[4])
        memory_length = max(1, int(self.genes[4] * 10) + 1)
        recent_history = self.history[-memory_length:]
        
        # Calculate opponent's recent COOPERATION rate (not defection!)
        cooperation_rate = sum(recent_history) / len(recent_history)
        
        # If opponent is mostly cooperating, reciprocate
        if cooperation_rate >= 0.5:
            return random.random() < self.genes[1]  # gene[1] = cooperate response
        
        # If opponent is mostly defecting, retaliate or forgive
        else:
            # Retaliate with probability gene[2]
            if random.random() < self.genes[2]:
                return UNDERCUT
            # Or forgive with probability gene[3]
            elif random.random() < self.genes[3]:
                return INVEST
            else:
                return UNDERCUT
```

```python

# ============================================================================
# PART 2: DEFINE THE OPPONENT ECOSYSTEM
# ============================================================================

# ============================================================================
# EXPANDED OPPONENT POOL
# ============================================================================

# Full pool of all possible opponents
full_opponent_pool = [
    # Cooperative agents
    AlwaysInvestAgent(),
    TitForTatAgent(),
    GrimTriggerAgent(),
    PavlovAgent(),
    TitForTwoTatsAgent(),
    GenerousTitForTatAgent(),
    AdaptiveAgent(),
    
    # Aggressive agents
    AlwaysUndercutAgent(),
    SuspiciousTitForTatAgent(),
    
    # Sophisticated agents
    GradualAgent(),
    ProberAgent(),
    
    # Majority-based agents
    HardMajorityAgent(),
    SoftMajorityAgent(),
    
    # Random agents with different cooperation levels
    RandomAgent(0.9),
    RandomAgent(0.7),
    RandomAgent(0.5),
    RandomAgent(0.3),
    RandomAgent(0.1),
]

print(f"\n📚 Full opponent library: {len(full_opponent_pool)} agents available")

# For initial evolution, randomly sample opponents
opponent_pool = random.sample(full_opponent_pool, 10)

print(f"🎲 Randomly selected {len(opponent_pool)} opponents for this evolution:")
for i, opp in enumerate(opponent_pool, 1):
    print(f"  {i:2d}. {opp.name:25s} - {opp.description}")
print("="*70)

```

```python

# ============================================================================
# PART 3: GENETIC ALGORITHM FUNCTIONS (YOUR TODO!)
# ============================================================================

def initialize_population(pop_size, seed_genes=None):
    """
    Create a well-spread initial population.
 
    Args:
        pop_size: Number of agents in the population
        seed_genes: Optional list of gene lists to inject (e.g. past winners)
 
    Returns:
        List of EvolvableAgent objects
    """
    # Find out how many genes an agent has instead of hardcoding 6
    n_genes = len(EvolvableAgent(name="probe").genes)
 
    population = []
    
    # Inject any seeded individuals first (copied so originals stay safe)
    for genes in (seed_genes or [])[:pop_size]:
        population.append(EvolvableAgent(genes=list(genes)))
 
    # Fill the remainder with stratified samples
    n_random = pop_size - len(population)
    if n_random > 0:
        columns = []
        for _ in range(n_genes):
            # One random point inside each of n_random equal slices of [0, 1]
            column = [(i + random.random()) / n_random for i in range(n_random)]
            # then shuffle so genes aren't correlated with each other
            random.shuffle(column)
            columns.append(column)
 
        for i in range(n_random):
            genes = [columns[g][i] for g in range(n_genes)]
            population.append(EvolvableAgent(genes=genes))
 
    return population


def evaluate_fitness(agent, opponents, num_rounds=50, games_per_opponent=1):
    """
    Average score per game for one agent against all opponents.
 
    Returns:
        Float: mean score per game (higher is better)
    """
    total_score = 0.0
    num_games = 0
 
    for opponent in opponents:
        if opponent is agent:          # never play yourself
            continue
        for _ in range(games_per_opponent):
            game = Game(agent, opponent, num_rounds=num_rounds)
            score, _ = game.play()
            total_score += score
            num_games += 1
 
    if num_games == 0:
        raise ValueError("evaluate_fitness: no valid opponents to play against")
 
    return total_score / num_games
 
 
def evaluate_population(population, opponents, num_rounds=50,
                        games_per_opponent=1, cache=None):
    """
    Evaluate every agent. Pass cache={} to reuse results for identical genes
    (deterministic games only).
 
    Returns:
        List of fitness floats parallel to population
    """
    fitnesses = []
    for agent in population:
        key = tuple(round(g, 8) for g in agent.genes)
        if cache is not None and key in cache:
            fitnesses.append(cache[key])
            continue
 
        fitness = evaluate_fitness(agent, opponents, num_rounds, games_per_opponent)
        if cache is not None:
            cache[key] = fitness
        fitnesses.append(fitness)
 
    return fitnesses


def tournament_selection(population, fitnesses, tournament_size=3,
                         selection_pressure=0.9):
    """
    Select len(population) parents via (probabilistic) tournament selection.
 
    Returns:
        List of NEW EvolvableAgent copies (genes are copied, never shared)
    """
    tournament_size = max(1, min(tournament_size, len(population)))
    selected = []
 
    for _ in range(len(population)):
        # Pick random contestants without replacement within a tournament
        contestants = random.sample(range(len(population)), tournament_size)
 
        # Rank contestants best to worst
        contestants.sort(key=lambda i: fitnesses[i], reverse=True)
 
        # Walk down the ranking and take each with probability p, else move on
        winner = contestants[-1]  
        for idx in contestants:
            if random.random() < selection_pressure:
                winner = idx
                break
 
        selected.append(EvolvableAgent(genes=list(population[winner].genes)))
 
    return selected


def crossover_and_mutate(parent1, parent2, mutation_rate=0.15,
                         mutation_std=0.2, crossover_mode="blend",
                         blend_alpha=0.25):
    """
    Create one child from two parents via crossover then Gaussian mutation.
 
    Returns:
        New EvolvableAgent
    """
    p1, p2 = parent1.genes, parent2.genes
    n = len(p1)
 
    # Crossover 
    if crossover_mode == "single_point":
        point = random.randint(1, n - 1)
        child_genes = list(p1[:point]) + list(p2[point:])
 
    elif crossover_mode == "uniform":
        child_genes = [random.choice((a, b)) for a, b in zip(p1, p2)]
 
    elif crossover_mode == "blend":
        child_genes = []
        for a, b in zip(p1, p2):
            lo, hi = min(a, b), max(a, b)
            spread = (hi - lo) * blend_alpha
            child_genes.append(random.uniform(lo - spread, hi + spread))
 
    else:
        raise ValueError(f"Unknown crossover_mode: {crossover_mode}")
 
    # Mutation (Gaussian noise, then clip into [0, 1])
    for i in range(n):
        if random.random() < mutation_rate:
            child_genes[i] += random.gauss(0, mutation_std)
        child_genes[i] = max(0.0, min(1.0, child_genes[i]))  # also clips blend overshoot
 
    return EvolvableAgent(genes=child_genes)

def gene_diversity(population):
    """Average per-gene standard deviation across the population (0 = clones)."""
    n_genes = len(population[0].genes)
    stds = [statistics.pstdev([a.genes[g] for a in population]) for g in range(n_genes)]
    return sum(stds) / n_genes

def mutation_scale(gen, generations, final_scale=0.4):
    """
    Linear anneal: start exploring (scale 1.0), end fine-tuning (final_scale).
    Multiply mutation_std by this value each generation.
    """
    progress = gen / max(1, generations - 1)
    return 1.0 - (1.0 - final_scale) * progress


def evolve(generations=25, pop_size=20, opponents=None, mutation_rate=0.15,
           elitism=2, mutation_std=0.2, tournament_size=3,
           selection_pressure=0.9, crossover_mode="blend",
           games_per_opponent=1, num_rounds=50,
           stagnation_limit=6, immigrant_frac=0.2, early_stop_patience=None,
           use_cache=False, seed=None, plot=True, return_history=False):
    """
    Run the genetic algorithm and return the best agent found.
 
    Key knobs:
        stagnation_limit:   gens without improvement before injecting immigrants
        immigrant_frac:     fraction of population replaced by fresh random agents
        early_stop_patience: stop after this many stagnant gens (None = never)
        use_cache:          reuse fitness for identical genes (deterministic Game only)
    """
    if seed is not None:
        random.seed(seed)
    if opponents is None:
        opponents = opponent_pool
 
    elitism = max(0, min(elitism, pop_size - 1))
    cache = {} if use_cache else None
 
    print(f"\n🧬 EVOLUTION STARTING")
    print(f"{'='*70}")
    print(f"Population: {pop_size} | Generations: {generations} | Mutation: {mutation_rate}")
    print(f"Opponents: {len(opponents)} | Elitism: {elitism} | Crossover: {crossover_mode}")
    print(f"{'='*70}\n")
 
    history = {"best": [], "avg": [], "best_so_far": [], "diversity": []}
 
    population = initialize_population(pop_size)
    best_overall_agent = None
    best_overall_fitness = -float("inf")
    stagnant_gens = 0
 
    for gen in range(generations):
        # ---- 1. Evaluate ----
        fitnesses = evaluate_population(population, opponents, num_rounds,
                                        games_per_opponent, cache)
 
        # ---- 2. Track statistics ----
        best_fitness = max(fitnesses)
        avg_fitness = sum(fitnesses) / len(fitnesses)
        diversity = gene_diversity(population)
        best_idx = fitnesses.index(best_fitness)
 
        # Sanity check: with elitism, best should rarely drop. Small dips are
        # normal if Game is stochastic; big ones suggest a bug.
        if elitism > 0 and gen > 0 and best_fitness < history["best"][-1] - 1e-9:
            print(f"   ⚠️  Best dipped {history['best'][-1]:.2f} -> {best_fitness:.2f} "
                  f"(expected only if games are noisy)")
 
        # ---- 3. Track best agent ever seen (copy genes!) ----
        if best_fitness > best_overall_fitness + 1e-9:
            best_overall_fitness = best_fitness
            best_overall_agent = EvolvableAgent(genes=list(population[best_idx].genes))
            stagnant_gens = 0
        else:
            stagnant_gens += 1
 
        history["best"].append(best_fitness)
        history["avg"].append(avg_fitness)
        history["best_so_far"].append(best_overall_fitness)
        history["diversity"].append(diversity)
 
        print(f"Gen {gen+1:3d}/{generations}: "
              f"Best={best_fitness:6.2f} | Avg={avg_fitness:6.2f} | "
              f"Div={diversity:.3f} | Stall={stagnant_gens} | "
              f"Genes={[f'{g:.2f}' for g in population[best_idx].genes]}")
 
        # ---- 4. Early stopping ----
        if early_stop_patience is not None and stagnant_gens >= early_stop_patience:
            print(f"\n⏹️  Early stop: no improvement for {stagnant_gens} generations.")
            break
 
        # ---- 5. Build next generation (skip after the final one) ----
        if gen < generations - 1:
            # Elitism: carry the top agents over unchanged
            ranked = sorted(range(len(fitnesses)), key=lambda i: fitnesses[i], reverse=True)
            next_population = [EvolvableAgent(genes=list(population[i].genes))
                               for i in ranked[:elitism]]
 
            # Random immigrants if we've been stuck (fresh genes escape local optima)
            if stagnant_gens >= stagnation_limit:
                n_imm = max(1, int(pop_size * immigrant_frac))
                n_imm = min(n_imm, pop_size - len(next_population))
                next_population += initialize_population(n_imm)
                print(f"   🌱 Stagnation: injected {n_imm} random immigrants")
                stagnant_gens = 0
 
            # Anneal mutation strength over the run
            std_now = mutation_std * mutation_scale(gen, generations)
 
            # Selection -> crossover -> mutation fills the remaining slots
            selected = tournament_selection(population, fitnesses,
                                            tournament_size, selection_pressure)
            while len(next_population) < pop_size:
                parent1, parent2 = random.sample(selected, 2)  # two different slots
                child = crossover_and_mutate(parent1, parent2, mutation_rate,
                                             std_now, crossover_mode)
                next_population.append(child)
 
            population = next_population
 
    print(f"\n{'='*70}")
    print(f"✅ EVOLUTION COMPLETE!")
    print(f"🏆 Best fitness achieved: {best_overall_fitness:.2f}")
    print(f"   Best genes: {[round(g, 3) for g in best_overall_agent.genes]}")
    print(f"{'='*70}\n")
 
    # ---- 6. Plot: fitness on top, diversity below ----
    if plot:
        xs = range(1, len(history["best"]) + 1)
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True,
                                       gridspec_kw={"height_ratios": [3, 1]})
        ax1.plot(xs, history["best"], label="Best (this gen)", color="green",
                 linewidth=2, marker="o")
        ax1.plot(xs, history["best_so_far"], label="Best so far", color="darkgreen",
                 linestyle="--")
        ax1.plot(xs, history["avg"], label="Average", color="blue",
                 linewidth=2, marker="s", alpha=0.7)
        ax1.set_ylabel("Fitness (Avg Score per Game)", fontsize=12)
        ax1.set_title("Evolution Progress", fontsize=14, fontweight="bold")
        ax1.legend(fontsize=11)
        ax1.grid(True, alpha=0.3)
 
        ax2.plot(xs, history["diversity"], color="purple", linewidth=2)
        ax2.set_xlabel("Generation", fontsize=12)
        ax2.set_ylabel("Gene diversity", fontsize=11)
        ax2.grid(True, alpha=0.3)
 
        plt.tight_layout()
        plt.show()
 
    if return_history:
        return best_overall_agent, history
    return best_overall_agent

```

```python

# ============================================================================
# PART 6: IMPROVE YOUR AGENT'S ACTION TAKING STRATEGY
# ============================================================================

# If the opponent defects after BOTH of us cooperated more often than this,
# it is acting "unprovoked" (random / noisy), not reacting to me.
UNPROVOKED_LIMIT = 0.22
 
 
class ImprovedEvolvableAgent(Agent):
    """
    10-gene agent. Every gene is in [0, 1]; traits() decodes them.
 
    Gene map
      0 open_coop       P(cooperate) on the very first move
      1 trust_coop      P(cooperate) while the opponent looks cooperative
      2 retaliation     P(punish) after I cooperated and was defected on
      3 forgiveness     P(offer cooperation) to break defection spirals and
                        after an opponent's FIRST-move defection
      4 memory          how many recent rounds I judge the opponent by (2-12)
      5 coop_threshold  cooperation rate needed to count as "cooperative"
      6 detect_threshold responsiveness below this = "opponent ignores me"
      7 exploit_prob    P(defect) against an unresponsive / random opponent
      8 endgame_len     how many final rounds to defect in (0-4)
      9 streak_forgive  P(forgive a lone defection after a long cooperation streak)
    """
 
    N_GENES = 10
 
    # Set this to your game's round count (e.g. ImprovedEvolvableAgent.ASSUMED_ROUNDS = 50)
    # to enable end-game defection. None = end-game defection disabled.
    ASSUMED_ROUNDS = None
 
    def __init__(self, genes=None, name="Improved Agent"):
        if genes is None:
            genes = [random.random() for _ in range(self.N_GENES)]
        # Pad / trim so an old 6-gene vector can never crash the agent
        genes = list(genes) + [random.random() for _ in range(self.N_GENES - len(genes))]
        self.genes = genes[:self.N_GENES]
        super().__init__(name, f"Genes: [{', '.join(f'{g:.2f}' for g in self.genes)}]")
        self.my_moves = []   # my own past moves (needed for the correlation test)
 
    # Gene decoding
    def traits(self):
        g = self.genes
        return {
            "open_coop": g[0],
            "trust_coop": g[1],
            "retaliation": g[2],
            "forgiveness": g[3],
            "memory": 2 + int(g[4] * 10),              # 2-12 rounds
            "coop_threshold": 0.5 + 0.4 * g[5],        # 0.50-0.90
            "detect_threshold": 0.05 + 0.40 * g[6],    # 0.05-0.45
            "exploit_prob": g[7],
            "endgame_len": min(4, int(g[8] * 5)),      # 0-4 rounds
            "streak_forgive": g[9],
        }
 
    def _horizon(self):
        """Total rounds in the game, if we can find out."""
        for attr in ("num_rounds", "total_rounds", "max_rounds"):
            val = getattr(self, attr, None)
            if isinstance(val, int) and val > 0:
                return val
        return self.ASSUMED_ROUNDS
 
    # UPGRADE 1 - detect random / unresponsive opponents
    def _responsiveness(self, opp, mine, window=30, min_samples=3):
        """
        Correlation-style test: does the opponent's NEXT move depend on
        MY move?   responsiveness = P(opp coops | I coop) - P(opp coops | I defect)
          ~ +1  -> reacts to me (Tit-for-Tat, Grim, Gradual...)
          ~  0  -> ignores me (Random, Always X)  -> cooperation is wasted
        Returns None until I have enough samples of BOTH of my move types.
        """
        pairs = list(zip(mine[:-1], opp[1:]))[-window:]   # (my move, their reply)
        after_c = [o for m, o in pairs if m]
        after_d = [o for m, o in pairs if not m]
        if len(after_c) < min_samples or len(after_d) < min_samples:
            return None
        rate_c = (sum(after_c) + 1) / (len(after_c) + 2)   # Laplace smoothing
        rate_d = (sum(after_d) + 1) / (len(after_d) + 2)
        return rate_c - rate_d
 
    def _unprovoked_rate(self, opp, mine, window=30, min_samples=6):
        """
        Backup detector for when I have (almost) never defected, so the test
        above has no data: after rounds where BOTH of us cooperated, how often
        does the opponent defect anyway? Reactive bots ~0, random bots high.
        """
        nxt = [opp[i + 1] for i in range(len(opp) - 1) if mine[i] and opp[i]][-window:]
        if len(nxt) < min_samples:
            return None
        return 1 - sum(nxt) / len(nxt)
 
    def _coop_streak(self, opp, mine, end):
        """Length of the mutual-cooperation streak ending at index `end`."""
        streak = 0
        for i in range(end, -1, -1):
            if mine[i] and opp[i]:
                streak += 1
            else:
                break
        return streak
 
    # Decision logic
    def choose_action(self) -> bool:
        if len(self.history) == 0:          # new game will wipe my own move log
            self.my_moves = []
 
        opp = [bool(x) for x in self.history]
        n = min(len(opp), len(self.my_moves))
        opp, mine = opp[:n], self.my_moves[:n]
 
        cooperate = self._decide(self.traits(), opp, mine, n)
        self.my_moves.append(bool(cooperate))
        return INVEST if cooperate else UNDERCUT
 
    def _decide(self, tr, opp, mine, n):
        # Opening move 
        if n == 0:
            return random.random() < tr["open_coop"]
 
        # End-game defection (opponent can't punish after the end) 
        horizon = self._horizon()
        if (horizon is not None and tr["endgame_len"] > 0
                and len(self.history) >= horizon - tr["endgame_len"]):
            return False
 
        # Random / unresponsive opponent = exploit 
        resp = self._responsiveness(opp, mine)
        if resp is not None:
            if resp < tr["detect_threshold"]:
                return random.random() >= tr["exploit_prob"]
        else:
            unprov = self._unprovoked_rate(opp, mine)
            if unprov is not None and unprov > UNPROVOKED_LIMIT:
                return random.random() >= tr["exploit_prob"]
 
        # Reactive play (opponent responds to me) 
        recent = opp[-tr["memory"]:]
        coop_rate = sum(recent) / len(recent)
        last_opp, last_me = opp[-1], mine[-1]
 
        # Opponent cooperated last round
        if last_opp:
            if coop_rate >= tr["coop_threshold"]:
                return random.random() < tr["trust_coop"]
            return random.random() < tr["trust_coop"] * coop_rate  # cautious
 
        # Opponent defected last round 
        # UPGRADE 2a: first-move amnesty (Suspicious Tit-for-Tat opens with D;
        # retaliating just starts an endless echo)
        if n == 1:
            return random.random() < tr["forgiveness"]
 
        # UPGRADE 2b: forgive a LONE defection after a long cooperation streak
        # (treat it as noise instead of starting a punishment spiral)
        if recent.count(False) == 1:
            streak = self._coop_streak(opp, mine, n - 2)
            if streak >= 4 and random.random() < tr["streak_forgive"]:
                return True
 
        if last_me:   # I cooperated and got betrayed = punish
            return random.random() >= tr["retaliation"]
 
        # Mutual defection = offer an olive branch to escape the spiral
        return random.random() < tr["forgiveness"]
 
    # Save / load
    def save_to_file(self, filename):
        data = {"name": self.name, "genes": self.genes,
                "traits": self.traits(), "description": self.description}
        with open(filename, "w") as f:
            json.dump(data, f, indent=2)
        print(f"✅ Agent saved to {filename}")
 
    @classmethod
    def load_from_file(cls, filename):
        with open(filename) as f:
            data = json.load(f)
        return cls(genes=data["genes"], name=data.get("name", "Loaded Agent"))
 
 
class AlwaysUndercutBot(Agent):
    def __init__(self):
        super().__init__("Always Undercut (train)", "Defects every round")
    def choose_action(self):
        return UNDERCUT
 
 
class BiasedRandom(Agent):
    def __init__(self, p):
        super().__init__(f"Biased Random ({p})", f"Cooperates with probability {p}")
        self.p = p
    def choose_action(self):
        return INVEST if random.random() < self.p else UNDERCUT
 
 
class NoisyTitForTat(Agent):
    """Tit-for-Tat that accidentally flips its move with probability `noise`."""
    def __init__(self, noise=0.1):
        super().__init__(f"Noisy TFT ({noise})", f"Tit-for-Tat with {noise:.0%} noise")
        self.noise = noise
    def choose_action(self):
        move = True if len(self.history) == 0 else bool(self.history[-1])
        if random.random() < self.noise:
            move = not move
        return INVEST if move else UNDERCUT
 
 
class JossBot(Agent):
    """Tit-for-Tat that sneakily defects when it would have cooperated."""
    def __init__(self, sneak=0.1):
        super().__init__(f"Joss ({sneak})", f"TFT with {sneak:.0%} sneak defections")
        self.sneak = sneak
    def choose_action(self):
        move = True if len(self.history) == 0 else bool(self.history[-1])
        if move and random.random() < self.sneak:
            move = False
        return INVEST if move else UNDERCUT
 
 
class AlternatorBot(Agent):
    def __init__(self):
        super().__init__("Alternator", "Alternates invest / undercut")
    def choose_action(self):
        return INVEST if len(self.history) % 2 == 0 else UNDERCUT
 
 
class ProberBot(Agent):
    """Opens D, C, C. If I never punished it, it defects forever; else TFT."""
    def __init__(self):
        super().__init__("Prober", "Tests whether you retaliate, then exploits")
    def choose_action(self):
        r = len(self.history)
        if r == 0:
            return UNDERCUT
        if r in (1, 2):
            return INVEST
        if self.history[1] and self.history[2]:     # I let it get away with it
            return UNDERCUT
        return INVEST if self.history[-1] else UNDERCUT
 
 
class HardMajorityBot(Agent):
    def __init__(self):
        super().__init__("Hard Majority", "Defects if you defected at least half the time")
    def choose_action(self):
        if len(self.history) == 0:
            return UNDERCUT
        defects = sum(1 for x in self.history if not x)
        return UNDERCUT if defects >= len(self.history) / 2 else INVEST
 
 
class NoisyPavlov(Agent):
    """Win-stay / lose-shift with random slips."""
    def __init__(self, noise=0.1):
        super().__init__(f"Noisy Pavlov ({noise})", f"Pavlov with {noise:.0%} noise")
        self.noise = noise
        self.moves = []
    def choose_action(self):
        if len(self.history) == 0:
            self.moves = []
            move = True
        else:
            last_me, last_opp = self.moves[-1], bool(self.history[-1])
            move = last_me if last_opp else (not last_me)
        if random.random() < self.noise:
            move = not move
        self.moves.append(move)
        return INVEST if move else UNDERCUT
 
 
def build_training_pool(base_pool):
    """Your original opponents + noisy / random / tricky extras (no duplicates)."""
    extras = [
        AlwaysUndercutBot(), NoisyTitForTat(0.05), NoisyTitForTat(0.15),
        JossBot(0.1), BiasedRandom(0.5), BiasedRandom(0.9),
        AlternatorBot(), ProberBot(), HardMajorityBot(), NoisyPavlov(0.1),
    ]
    existing = {a.name for a in base_pool}
    return list(base_pool) + [a for a in extras if a.name not in existing]
 
 
# Opponents the training NEVER sees - used at the end to catch overfitting
def build_holdout_pool():
    return [NoisyTitForTat(0.25), BiasedRandom(0.2), BiasedRandom(0.6),
            JossBot(0.2), NoisyPavlov(0.2)]

def _play_match(agent, opponent, num_rounds):
    """Returns (my_score, opp_score). opp_score is None if play() doesn't give it."""
    result = Game(agent, opponent, num_rounds=num_rounds).play()
    my_score = result[0]
    opp_score = result[1] if len(result) > 1 and isinstance(result[1], (int, float)) else None
    return my_score, opp_score
 
 
def evaluate_fitness(agent, opponents, num_rounds=50, games_per_opponent=1,
                     win_bonus=40, loss_penalty=60,
                     worst_case_weight=0.3, worst_k=3):
    per_opponent = []
    for opponent in opponents:
        if opponent is agent:
            continue
        total = 0.0
        for _ in range(games_per_opponent):
            mine, theirs = _play_match(agent, opponent, num_rounds)
            total += mine
            if theirs is not None:
                if mine > theirs:
                    total += win_bonus
                elif mine < theirs:
                    total -= loss_penalty
        per_opponent.append(total / games_per_opponent)
 
    if not per_opponent:
        raise ValueError("evaluate_fitness: no valid opponents")
 
    overall = sum(per_opponent) / len(per_opponent)
    k = min(worst_k, len(per_opponent))
    worst = sum(sorted(per_opponent)[:k]) / k
    return (1 - worst_case_weight) * overall + worst_case_weight * worst
 
 
# %% CELL D - Train the new agent
ROUNDS = 50   # <-- set to the SAME num_rounds your benchmark Game uses
 
# Every earlier GA function builds agents with the name EvolvableAgent, so
# pointing that name at the new class makes the whole pipeline evolve it.
EvolvableAgent = ImprovedEvolvableAgent
ImprovedEvolvableAgent.ASSUMED_ROUNDS = ROUNDS   # enables end-game defection
 
training_pool = build_training_pool(opponent_pool)
print(f"Training against {len(training_pool)} opponents")
 
best_agent, history = evolve(
    generations=60,
    pop_size=30,
    opponents=training_pool,
    num_rounds=ROUNDS,
    games_per_opponent=3,        # raise to 5 for the final long run
    mutation_rate=0.2,
    mutation_std=0.2,
    elitism=3,
    tournament_size=4,
    selection_pressure=0.9,
    crossover_mode="blend",
    stagnation_limit=8,
    immigrant_frac=0.2,
    return_history=True,
)
 
print("\nEvolved traits:")
for k, v in ImprovedEvolvableAgent(genes=best_agent.genes).traits().items():
    print(f"  {k:16s} {v:.3f}" if isinstance(v, float) else f"  {k:16s} {v}")
 
 
# %% CELL E - Benchmark the champion (incl. opponents it never trained on)
def quick_benchmark(agent, pool, num_rounds=ROUNDS, games=5):
    wins = losses = ties = 0
    for opp in pool:
        mine = theirs = 0.0
        for _ in range(games):
            m, o = _play_match(agent, opp, num_rounds)
            mine += m
            theirs += (o if o is not None else 0)
        mine /= games
        theirs /= games
        if mine > theirs + 1e-9:
            tag, wins = "WON ", wins + 1
        elif mine < theirs - 1e-9:
            tag, losses = "LOST", losses + 1
        else:
            tag, ties = "TIED", ties + 1
        print(f"vs {opp.name:28s}: {mine:7.1f} - {theirs:7.1f}  {tag}")
    print(f"\nRecord: {wins} W / {losses} L / {ties} T")
 
 
champion = ImprovedEvolvableAgent(genes=list(best_agent.genes), name="Champion")
 
print("=== Original benchmark pool ===")
quick_benchmark(champion, opponent_pool)
 
print("\n=== Extra training opponents ===")
quick_benchmark(champion, [a for a in training_pool if a not in opponent_pool])
 
print("\n=== HOLD-OUT (never seen in training) ===")
quick_benchmark(champion, build_holdout_pool())
 
champion.save_to_file("champion_agent.json")
```

```python
# ============================================================================
# PART 7: EVOLVE YOUR IMPROVED AGENT
# ============================================================================


print("\n" + "="*70)
print("🔧 EVOLVING YOUR IMPROVED AGENT!")
print("="*70)


def evolve_improved(generations=25, pop_size=20, opponents=None, mutation_rate=0.15, elitism=2):
    """
    Evolve using the ImprovedEvolvableAgent class
    (Same as evolve() but uses ImprovedEvolvableAgent instead of EvolvableAgent)
    """
    if opponents is None:
        opponents = opponent_pool
    
    print(f"\n🧬 EVOLUTION STARTING (Improved Agent)")
    print(f"{'='*70}")
    print(f"Population: {pop_size} | Generations: {generations} | Mutation: {mutation_rate}")
    print(f"Opponents: {len(opponents)} | Elitism: {elitism}")
    print(f"{'='*70}\n")
    
    best_fitness_history = []
    avg_fitness_history = []
    
    # Initialize population with ImprovedEvolvableAgent
    population = [ImprovedEvolvableAgent(name=f"Agent_{i}") for i in range(pop_size)]
    
    best_overall_agent = None
    best_overall_fitness = -float('inf')
    
    for gen in range(generations):
        # Evaluate fitness
        fitnesses = []
        for agent in population:
            fitness = evaluate_fitness(agent, opponents, num_rounds=50)
            fitnesses.append(fitness)
        
        # Track best
        best_fitness = max(fitnesses)
        avg_fitness = sum(fitnesses) / len(fitnesses)
        
        # Update best overall
        best_idx = fitnesses.index(best_fitness)
        if best_fitness > best_overall_fitness:
            best_overall_fitness = best_fitness
            best_overall_agent = ImprovedEvolvableAgent(genes=population[best_idx].genes.copy())
        
        best_fitness_history.append(best_overall_fitness)
        avg_fitness_history.append(avg_fitness)
        
        print(f"Gen {gen+1:3d}/{generations}: "
              f"Best={best_overall_fitness:6.2f} | Avg={avg_fitness:6.2f} | "
              f"Genes={[f'{g:.2f}' for g in population[best_idx].genes]}")
        
        # Create next generation
        if gen < generations - 1:
            sorted_indices = sorted(range(len(fitnesses)), 
                                  key=lambda i: fitnesses[i], 
                                  reverse=True)
            
            # Elitism
            next_population = []
            for i in range(elitism):
                elite_genes = population[sorted_indices[i]].genes.copy()
                next_population.append(ImprovedEvolvableAgent(genes=elite_genes))
            
            # Selection
            selected = tournament_selection(population, fitnesses)
            
            # Crossover and mutation
            while len(next_population) < pop_size:
                parent1, parent2 = random.sample(selected, 2)
                child = crossover_and_mutate(parent1, parent2, mutation_rate)
                # Convert to ImprovedEvolvableAgent
                next_population.append(ImprovedEvolvableAgent(genes=child.genes))
            
            population = next_population
    
    print(f"\n{'='*70}")
    print(f"✅ EVOLUTION COMPLETE!")
    print(f"🏆 Best fitness achieved: {best_overall_fitness:.2f}")
    print(f"{'='*70}\n")
    
    # Plot
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, generations+1), best_fitness_history, 
             label='Best Fitness', linewidth=2, color='green', marker='o')
    plt.plot(range(1, generations+1), avg_fitness_history, 
             label='Average Fitness', linewidth=2, color='blue', marker='s', alpha=0.7)
    plt.xlabel('Generation', fontsize=12)
    plt.ylabel('Fitness (Avg Score per Game)', fontsize=12)
    plt.title('Improved Agent Evolution Progress', fontsize=14, fontweight='bold')
    plt.legend(fontsize=11)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
    
    return best_overall_agent


# TODO: Experiment with these parameters to find the best improved agent!
my_improved_agent = evolve_improved(
    generations=100,      # Try: 20, 30, 50
    pop_size=30,         # Try: 15, 20, 30
    opponents=opponent_pool,
    mutation_rate=0.1,   # Try: 0.1, 0.15, 0.2
    elitism=2            # Keep top 2 agents each generation
)

print(f"\n🎉 Your improved agent's final genes:")
print(f"   {my_improved_agent.genes}")
```
